## 9. Gradient Descent

Gradient descent repeatedly adjusts parameters in a direction that reduces loss.

For our linear model and MSE loss, the gradients are:

\[
\frac{\partial \text{MSE}}{\partial w}
= -\frac{2}{n}\sum_{i=1}^{n}x_i(y_i-\hat{y}_i)
\]

\[
\frac{\partial \text{MSE}}{\partial b}
= -\frac{2}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)
\]

The update rule is:

\[
w \leftarrow w - \alpha\frac{\partial \text{MSE}}{\partial w}
\]

\[
b \leftarrow b - \alpha\frac{\partial \text{MSE}}{\partial b}
\]

`alpha` (the learning rate) controls the size of each update.

You don't need to memorize these formulas immediately. Implement them and observe the effect of the updates.
